## ASOM-Aligned Threat Hunting Platform

This notebook provides a no-code interface for ASOM (Analytic Scheme of Maneuver) aligned threat hunting.

Each strategy answers a specific CCIR (Commander's Critical Information Requirement) using up to 3 sophistication levels:
- **Level 1**: Rule-based detection (watchlists, correlation rules)
- **Level 2**: Statistical baseline detection (anomaly scoring)
- **Level 3**: Machine learning detection (supervised/unsupervised ML)

**How to use:**
1. Run the cell below to initialize the dashboard
2. Select a hunt strategy from the dropdown (organized by CCIR)
3. Choose your data table
4. Map the required columns
5. Click "Run Analysis" to execute
6. Review results with severity filtering and ML insights


```python
# Import the dashboard and ASOM-aligned strategies
from app import WatsonDashboard
from strategies import get_all_strategies

# Get all available ASOM strategies
strategies = get_all_strategies()

# Display information about available strategies
print(f"\n{'='*80}")
print(f"Available ASOM-Aligned Strategies: {len(strategies)}")
print(f"{'='*80}\n")
for strategy in strategies:
    print(f"📊 CCIR {strategy.ccir}: {strategy.name}")
    print(f"   Tactic: {strategy._get_tactic()}")
    print(f"   Question: {strategy._get_ccir_question()}")
    print(f"   Sophistication Levels: Rule-Based, Statistical, Machine Learning")
    print()

# Create and display the dashboard
print(f"{'='*80}")
print("Initializing 221B Dashboard...")
print(f"{'='*80}\n")
dashboard = WatsonDashboard(strategies)
dashboard.display()
```
